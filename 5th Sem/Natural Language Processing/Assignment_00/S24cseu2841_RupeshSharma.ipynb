```python
def clean_text(text):

    # 1. Lowercase
    text = text.lower()

    # 2. Handle contractions / shorthand
    text = text.replace("can't", "cannot")
    text = text.replace("can’t", "cannot")
    text = text.replace("i've", "i have")
    text = text.replace("it's", "it is")
    text = text.replace("u're", "you are")
    text = text.replace("c'mon", "come on")
    text = text.replace("<3", "love")
    text = text.replace("&", "and")

    # 3. Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # 4. Reduce duplicate letters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # 5. Remove special characters / punctuation
    text = re.sub(r'[^a-z0-9\s]', ' ', text)

    # 6. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
import nltk

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.corpus import brown
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

import matplotlib.pyplot as plt
```

```python
print(brown.categories())
```

```python
brown_words = brown.words(categories='news')

text = ' '.join(brown_words)

print(text[:1000])
```

```python
tokens = word_tokenize(text)

print(tokens[:50])
```

```python
tokens = [
    word.lower()
    for word in tokens
    if word.isalpha()
]

print(tokens[:50])
print("Total words:", len(tokens))
```

```python
freq_dist = FreqDist(tokens)

print(freq_dist.most_common(20))
```

```python
top20 = freq_dist.most_common(20)

words = [item[0] for item in top20]
counts = [item[1] for item in top20]

plt.figure(figsize=(12, 6))

plt.bar(words, counts)

plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top 20 Words in Brown Corpus - News")

plt.xticks(rotation=45)

plt.show()
```

```python
from wordcloud import WordCloud
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in tokens]

print(stemmed_words[:50])
```

```python
stem_freq = FreqDist(stemmed_words)

print(stem_freq.most_common(20))
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in tokens
]

print(lemmatized_words[:50])
```

```python
lemma_freq = FreqDist(lemmatized_words)

print(lemma_freq.most_common(20))
```

```python
common_words = [
    word
    for word, count in freq_dist.most_common(15)
]

print(common_words)
```

```python
import pandas as pd

comparison = []

for word in common_words:

    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)

    comparison.append({
        "Original Word": word,
        "Original Frequency": freq_dist[word],
        "Stem": stem,
        "Stem Frequency": stem_freq[stem],
        "Lemma": lemma,
        "Lemma Frequency": lemma_freq[lemma]
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
import numpy as np

x = np.arange(len(comparison_df))

width = 0.35

plt.figure(figsize=(14, 7))

plt.bar(
    x - width/2,
    comparison_df["Stem Frequency"],
    width,
    label="Stemming"
)

plt.bar(
    x + width/2,
    comparison_df["Lemma Frequency"],
    width,
    label="Lemmatization"
)

plt.xlabel("Words")
plt.ylabel("Frequency")

plt.title("Stemming vs Lemmatization")

plt.xticks(
    x,
    comparison_df["Original Word"],
    rotation=45
)

plt.legend()

plt.show()
```
