# Q1. Sentence Tokenization, Word Tokenization and Stop-word Removal

First, I load the Brown Corpus and select the first 20 sentences. Then I perform sentence and word tokenization, convert the words to lowercase, and remove punctuation and stop words.

```python
import string
from nltk.corpus import brown, stopwords
from nltk.tokenize import sent_tokenize, word_tokenize

stop_words = set(stopwords.words('english'))

# Taking first 20 sentences from Brown Corpus
sentences = brown.sents()[:20]

print('Number of sentences selected:', len(sentences))
```

```python
# Display original and cleaned tokens for first 5 sentences
for i, sentence in enumerate(sentences[:5], start=1):
    sentence_text = ' '.join(sentence)

    # Sentence tokenization
    sentence_tokens = sent_tokenize(sentence_text)

    # Word tokenization
    word_tokens = word_tokenize(sentence_text)

    # Lowercase
    lower_tokens = [word.lower() for word in word_tokens]

    # Remove punctuation and stop words
    cleaned_tokens = [
        word for word in lower_tokens
        if word.isalpha() and word not in stop_words
    ]

    print('\n' + '=' * 60)
    print('Sentence', i)
    print('=' * 60)
    print('Original Sentence:', sentence_text)
    print('Sentence Tokens:', sentence_tokens)
    print('Word Tokens:', word_tokens)
    print('Cleaned Tokens:', cleaned_tokens)
```

# Q2. Stemming and Lemmatization

Here I extract alphabetic words from the Brown Corpus and compare Porter stemming with WordNet lemmatization. I have also included the required example words.

```python
import pandas as pd
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Get alphabetic words from Brown Corpus
brown_words = [
    word.lower() for word in brown.words() if word.isalpha()
]

# At least 30 words
sample_words = brown_words[:30]

# Required words from the question
sample_words += ['playing', 'studies', 'running', 'better', 'cars']

def get_pos(word):
    # POS tags for the required examples
    if word in ['playing', 'running']:
        return wordnet.VERB
    elif word == 'studies':
        return wordnet.NOUN
    elif word == 'better':
        return wordnet.ADJ
    else:
        return wordnet.NOUN

data = []

for word in sample_words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word, pos=get_pos(word))
    data.append([word, stem, lemma])

df = pd.DataFrame(
    data,
    columns=['Original Word', 'Stemmed', 'Lemmatized']
)

display(df)
```

### Observation

Stemming usually cuts a word to a root-like form, while lemmatization tries to give the proper dictionary form. For example, `studies` may become `studi` after stemming but `study` after lemmatization.

# Q3. Basic Spelling Correction

For spelling correction, I use the NLTK Words Corpus as the vocabulary and Python's `difflib` for approximate string matching.

```python
import difflib
import re
from nltk.corpus import words

# Load vocabulary from NLTK Words Corpus
word_list = sorted(set(
    word.lower() for word in words.words() if word.isalpha()
))

# 10 intentionally misspelled words
misspelled_words = [
    'computr',
    'langauge',
    'machien',
    'algoritm',
    'progrmming',
    'pyhton',
    'naturall',
    'inteligence',
    'lernning',
    'developement'
]

def suggest_corrections(word):
    return difflib.get_close_matches(
        word.lower(), word_list, n=3, cutoff=0.6
    )

print('Misspelled words and suggestions')
print('-' * 55)

for word in misspelled_words:
    print(f'{word:20} -> {suggest_corrections(word)}')
```

```python
# Function to correct a complete sentence
def correct_word(word):
    suggestions = suggest_corrections(word)
    if suggestions:
        return suggestions[0]
    return word

def correct_sentence(sentence):
    words_in_sentence = re.findall(r'[A-Za-z]+', sentence)
    corrected = [correct_word(word) for word in words_in_sentence]
    return ' '.join(corrected)

# Five example sentences
example_sentences = [
    'I love progrmming in Pyhton.',
    'Naturall langauge processing is interesting.',
    'Machien lernning is a part of inteligence systems.',
    'The algoritm needs developement.',
    'I am lernning natural langauge processing.'
]

print('Corrected sentences')
print('=' * 60)

for sentence in example_sentences:
    print('\nOriginal :', sentence)
    print('Corrected:', correct_sentence(sentence))
```

# Q4. Unigram, Bigram and Trigram Analysis

I preprocess the Brown Corpus by converting words to lowercase and keeping only alphabetic tokens. Then I generate unigrams, bigrams and trigrams and calculate their frequencies.

```python
from nltk.util import ngrams
from collections import Counter

# Preprocess Brown Corpus
tokens = [
    word.lower() for word in brown.words() if word.isalpha()
]

print('Total preprocessed tokens:', len(tokens))
print('First 30 tokens:')
print(tokens[:30])
```

```python
# Generate n-grams
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

# Count frequencies
unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print('Number of unigrams:', len(unigrams))
print('Number of bigrams:', len(bigrams))
print('Number of trigrams:', len(trigrams))
```

```python
# Top 10 unigrams
top_unigrams = pd.DataFrame(
    unigram_freq.most_common(10),
    columns=['Unigram', 'Frequency']
)

# Top 10 bigrams
top_bigrams = pd.DataFrame(
    bigram_freq.most_common(10),
    columns=['Bigram', 'Frequency']
)

# Top 10 trigrams
top_trigrams = pd.DataFrame(
    trigram_freq.most_common(10),
    columns=['Trigram', 'Frequency']
)

print('Top 10 Unigrams')
display(top_unigrams)

print('Top 10 Bigrams')
display(top_bigrams)

print('Top 10 Trigrams')
display(top_trigrams)
```
